## 測試一（原始資料：4 人、折價券 100 元）

```python
# 原始資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱（去除頭尾空白並轉為小寫）
clean_event_name = event_name.strip().lower()

# 2. 清理餐費文字並轉換型別（保留原始 meal_amount）
clean_meal_str = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_price = float(clean_meal_str)
people_count = int(people)

# 3. 分步計算費用（先以原始餐費算服務費，再扣除折價券）
service_fee = meal_price * service_rate
total_amount = meal_price + service_fee - discount
per_person_cost = total_amount / people_count

# 4. 判斷每人費用是否大於每人預算（布林值）
is_over_budget = per_person_cost > budget_per_person

# 5. 輸出費用摘要
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_price:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
