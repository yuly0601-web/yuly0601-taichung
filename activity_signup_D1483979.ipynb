## 題目說明
本程式用於整理「桌遊體驗」與「電影欣賞」兩場活動的原始報名文字。透過字串處理統一格式後，利用 `list`、`tuple`、`set`、`dict` 進行去重、跨場次名單比對（交集、聯集、差集），並計算各場次與總計的應收費用。

## 測試一：原始資料

```python
# ================= 原始資料（測試一） =================
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# ================= 1. 讀取活動名稱與整理字串 =================
board_name, movie_name = activity_names

board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# ================= 2. 轉成集合去重與名單比較 =================
board_set = set(board_list)
movie_set = set(movie_list)

common_set = board_set & movie_set       # 交集：兩場都參加
all_set = board_set | movie_set          # 聯集：至少參加一場
only_board_set = board_set - movie_set   # 差集：只參加桌遊
only_movie_set = movie_set - board_set   # 差集：只參加電影

# ================= 3. 排序名單與切片預覽 =================
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
common_sorted = sorted(common_set)
only_board_sorted = sorted(only_board_set)
only_movie_sorted = sorted(only_movie_set)
all_sorted = sorted(all_set)

preview_list = all_sorted[:2]

# ================= 4. 建立摘要字典並計算費用 =================
board_count = len(board_set)
movie_count = len(movie_set)
board_fee = unit_fees[board_name]
movie_fee = unit_fees[movie_name]

summary = {
    "board_count": board_count,
    "movie_count": movie_count,
    "common_count": len(common_set),
    "total_unique": len(all_set),
    "total_registrations": board_count + movie_count,
    "board_revenue": board_count * board_fee,
    "movie_revenue": movie_count * movie_fee,
    "total_revenue": (board_count * board_fee) + (movie_count * movie_fee),
}

# ================= 5. 輸出完整摘要 =================
print("HW09 活動報名摘要")
print(f"活動：{board_name}，每人 {board_fee:.2f} 元")
print(f"活動：{movie_name}，每人 {movie_fee:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{common_sorted}")
print(f"共同人數：{summary['common_count']} 人")
print(f"只參加桌遊者：{only_board_sorted}")
print(f"只參加電影者：{only_movie_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['total_unique']} 人")
print(f"報名人次：{summary['total_registrations']} 人次")
print(f"通知名單前兩位：{preview_list}")
print(f"桌遊應收：{summary['board_revenue']:.2f} 元")
print(f"電影應收：{summary['movie_revenue']:.2f} 元")
print(f"總應收：{summary['total_revenue']:.2f} 元")
```

## 資料型別用途說明
* **`tuple` (`activity_names`)**：保存固定不變的兩個活動名稱（桌遊體驗、電影欣賞）。
* **`list` (`board_list`、`movie_list` 及排序後名單)**：保存清理空白與大小寫後的原始報名順序（含重複項），以及排序後的輸出名單。
* **`set` (`board_set`、`movie_set` 等)**：自動去除同一活動中的重複報名者，並執行交集、聯集與差集運算。
* **`dict` (`unit_fees`、`summary`)**：查詢各活動每人單價，以及集中儲存統計後的各項人數、人次與應收金額。
